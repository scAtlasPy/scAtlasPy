## Create a Small Synthetic QC Table

The example below creates a tiny synthetic cell-level summary table. Real
tutorials can replace this section with an actual scAtlasPy analysis workflow.

```python
rng = np.random.default_rng(7)
n_cells = 120

qc = pd.DataFrame(
    {
        "cell_id": [f"cell_{i:03d}" for i in range(n_cells)],
        "n_genes": rng.poisson(lam=1800, size=n_cells),
        "total_counts": rng.gamma(shape=6.0, scale=1200.0, size=n_cells).astype(int),
        "pct_mito": rng.beta(a=2.0, b=18.0, size=n_cells) * 100,
    }
)

qc.head()
```

## Summarize QC Metrics

Notebook outputs are included in the generated HTML page, so readers can
inspect the result without rerunning the code.

```python
qc[["n_genes", "total_counts", "pct_mito"]].describe().round(2)
```

## Plot the Distribution

Figures produced by matplotlib are also embedded in the documentation build.

```python
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.scatter(qc["total_counts"], qc["n_genes"], c=qc["pct_mito"], cmap="viridis", s=28)
ax.set_xlabel("Total counts")
ax.set_ylabel("Detected genes")
ax.set_title("Synthetic cell-level QC metrics")
fig.tight_layout()
```
